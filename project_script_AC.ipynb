```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
puncta = pd.read_csv('/Users/adelinechoi/Documents/intro2python/mock_RNApuncta _quant.csv')
puncta
```

```python
## Two Independent Sample T-Tests for Genotype Comparisons

puncta = puncta.rename(columns={
    'Avg # puncta per ROI (GFP)': 'D1R',
    'Avg # puncta per ROI (Cy50)': 'D2R'
})

# Separate data by genotype
WT = puncta[puncta['Genotype'] == 'WT']
HET = puncta[puncta['Genotype'] == 'HET']

# Welch's t-test for D1R and D2R between WT and HET
t_d1r, p_d1r = stats.ttest_ind(WT['D1R'], HET['D1R'], equal_var=False)
t_d2r, p_d2r = stats.ttest_ind(WT['D2R'], HET['D2R'], equal_var=False)

print("D1R WT vs HET:")
print("  t =", t_d1r)
print("  p =", p_d1r)

print("\nD2R WT vs HET:")
print("  t =", t_d2r)
print("  p =", p_d2r)
```
